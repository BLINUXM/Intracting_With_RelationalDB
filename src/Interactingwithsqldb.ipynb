```python
%%sql
SELECT 
    title AS film_title,
    length AS film_length,
    release_year AS film_release_year 
FROM film 
LIMIT 10;
```

```python
%%sql 
SELECT 
    city,
    address,
    district
FROM address AS addr
JOIN city AS cty 
ON addr.city_id = cty.city_id
LIMIT 10;
```

```python
%%sql 
SELECT  
    country,
    city,
    address
FROM country ctry
JOIN city cty 
ON ctry.country_id = cty.country_id 
JOIN address addr 
ON cty.city_id = addr.city_id
JOIN staff stff
ON addr.address_id = stff.address_id
WHERE stff.active;
```

```python
%%sql 
SELECT
    rating,
    COUNT(*) AS film_count
FROM film 
GROUP BY rating; 
```

```python
%%sql
SELECT 
    name AS category_name,
    COUNT(film.film_id) AS film_counts
FROM category AS cat
JOIN film_category 
ON cat.category_id = film_category.category_id
JOIN film 
ON film_category.film_id = film.film_id
GROUP BY cat.name
ORDER BY film_counts;
```
